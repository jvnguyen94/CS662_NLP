#### deserialization script

I then wrote the Python script `deserialization.py`. 

This script takes in an `.xml.gz` file, unzips it (gzip library), parses it into an XML tree structure (through lxml etree library), and uses XPath to search for the `<DOC>` tags with type="story" and grabs the text of the article, residing under the `<TEXT>` then `<P>`tags. I then append all that data into the `deserialized.txt` file.

#### Running deserialization script
The parsing is handled by the `run_deserialization.sh` script. 
It passes the `xml.gz` file as a system variable into the `deserialization.py` script and append the information to the `deserialized.txt` file.

After I parsed each of the `xml.gz` files, I ran a line count to check the output of the deserialization. It looks like I was able to append data from each of the files. I had also done a rough check by eye just to make sure I was grabbing the correct paragraphs through the `<P>` tags and that they were nested under the correct structure.
